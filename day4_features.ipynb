```python
import pandas as pd
df = pd.read_csv("data/clean.csv")
```

```python
df["sent_at"] = pd.to_datetime(df["sent_at"], errors = "coerce", utc = True, format= "mixed")
```

```python
df["hour"] = df["sent_at"].dt.hour
```

```python
df["is_weekend"] = df["sent_at"].dt.dayofweek >= 5
```

```python
df.head()
```

```python
df = pd.get_dummies(df, columns = ["network_provider"])
```

```python
df.head()
```

```python
df["parts_bucket"] = pd.cut(df["message_parts"], bins=[0, 1, 2, 3])
print(df["parts_bucket"].isna().sum())
```
